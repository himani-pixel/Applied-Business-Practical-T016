```python
import pandas as pd
import numpy as np
from scipy import stats

# Load Dataset
df = pd.read_csv("blood_donation_registry_ml_ready.csv")

age = df["age"].dropna()

mean_age = age.mean()
std_age = age.std()
n = len(age)

confidence_levels = [0.90, 0.95, 0.99]

print("CONFIDENCE INTERVAL ESTIMATION")
print("Mean Age:", mean_age)

for confidence in confidence_levels:

    standard_error = std_age / np.sqrt(n)

    margin_error = stats.t.ppf(
        (1 + confidence) / 2,
        n - 1
    ) * standard_error

    lower = mean_age - margin_error
    upper = mean_age + margin_error

    print("\nConfidence Level:", confidence * 100, "%")
    print("Lower Limit:", lower)
    print("Upper Limit:", upper)
```
