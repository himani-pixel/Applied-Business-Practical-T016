```python
from scipy.optimize import linprog

# Objective function
# We use negative values because linprog performs minimization
c = [-40, -60]

# Constraints
A = [
    [2, 3],
    [1, 2]
]

b = [60, 40]

# Solve Optimization Problem
result = linprog(
    c,
    A_ub=A,
    b_ub=b,
    bounds=[(0, None), (0, None)],
    method="highs"
)

print("OPTIMIZATION MODELING")

if result.success:

    standard_campaigns = result.x[0]
    targeted_campaigns = result.x[1]

    maximum_value = -result.fun

    print("Standard Campaigns:", standard_campaigns)
    print("Targeted Campaigns:", targeted_campaigns)
    print("Maximum Expected Responses:", maximum_value)

else:
    print("No optimal solution found.")
```
