```python
import pandas as pd

df = pd.read_csv("blood_donation_registry_ml_ready.csv")

print("FIRST 5 RECORDS")
print(df.head())

print("\nLAST 5 RECORDS")
print(df.tail())

print("\nSHAPE")
print(df.shape)

print("\nCOLUMN NAMES")
print(df.columns)

print("\nDATA TYPES")
print(df.dtypes)

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

print("\nUNIQUE VALUES")
print(df.nunique())

print("\nBASIC STATISTICS")
print(df.describe())
```

```python
import matplotlib.pyplot as plt

df["blood_type"].value_counts().plot(kind="bar")

plt.title("Blood Type Distribution")
plt.xlabel("Blood Type")
plt.ylabel("Number of Donors")
plt.show()
```

```python
df["region"].value_counts().plot(kind="bar")

plt.title("Donors by Region")
plt.xlabel("Region")
plt.ylabel("Number of Donors")
plt.show()
```

```python
plt.hist(df["age"], bins=10)

plt.title("Age Distribution of Donors")
plt.xlabel("Age")
plt.ylabel("Number of Donors")
plt.show()
```

```python
plt.hist(df["donation_count_last_12m"], bins=10)

plt.title("Donation Count in Last 12 Months")
plt.xlabel("Number of Donations")
plt.ylabel("Number of Donors")
plt.show()
```

```python
plt.boxplot(df["bmi"].dropna())

plt.title("BMI Distribution of Donors")
plt.ylabel("BMI")
plt.show()
```

```python
df["is_regular_donor"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Regular vs Non-Regular Donors")
plt.ylabel("")
plt.show()
```

```python
plt.scatter(
    df["age"],
    df["lifetime_donation_count"]
)

plt.title("Age vs Lifetime Donation Count")
plt.xlabel("Age")
plt.ylabel("Lifetime Donation Count")
plt.show()
```
