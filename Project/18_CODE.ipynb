```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("blood_donation_registry_ml_ready.csv")

# 1. Probability
total_donors = len(df)
donated = df["donated_next_6m"].sum()

probability = donated / total_donors

print("PROBABILITY")
print("Total Donors:", total_donors)
print("Donated in Next 6 Months:", donated)
print("Probability:", probability)
print("Probability (%):", probability * 100)


# 2. Generate Random Values

np.random.seed(42)

# Normal Distribution - Age
normal_data = np.random.normal(
    df["age"].mean(),
    df["age"].std(),
    1000
)

# Uniform Distribution - BMI
uniform_data = np.random.uniform(
    df["bmi"].min(),
    df["bmi"].max(),
    1000
)

# Exponential Distribution - Recency
exponential_data = np.random.exponential(
    df["recency_days"].mean(),
    1000
)


# 3. Normal Distribution

plt.hist(normal_data, bins=30)
plt.title("Normal Distribution of Donor Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()


# 4. Uniform Distribution

plt.hist(uniform_data, bins=30)
plt.title("Uniform Distribution of BMI")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.show()


# 5. Exponential Distribution

plt.hist(exponential_data, bins=30)
plt.title("Exponential Distribution of Recency")
plt.xlabel("Recency Days")
plt.ylabel("Frequency")
plt.show()
```
