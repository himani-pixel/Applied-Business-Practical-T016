```python
import pandas as pd
from scipy.stats import ttest_1samp

# Load Dataset
df = pd.read_csv("blood_donation_registry_ml_ready.csv")

# Select donor age
age = df["age"].dropna()

# Hypothesized mean
hypothesized_mean = 45

# One-sample t-test
t_statistic, p_value = ttest_1samp(age, hypothesized_mean)

print("HYPOTHESIS TESTING")
print("Sample Mean:", age.mean())
print("Hypothesized Mean:", hypothesized_mean)
print("T-Statistic:", t_statistic)
print("P-Value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```
