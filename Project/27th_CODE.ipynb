```python
import pandas as pd
import matplotlib.pyplot as plt

# Experimental Data
data = {
    "Campaign": [
        "Standard", "Standard", "Standard",
        "Personalized", "Personalized", "Personalized"
    ],
    "Incentive": [
        "Low", "Medium", "High",
        "Low", "Medium", "High"
    ],
    "Response_Rate": [
        30, 40, 50,
        35, 48, 60
    ]
}

df = pd.DataFrame(data)

print("EXPERIMENTAL DESIGN")
print(df)

# Plot
plt.bar(
    df["Campaign"] + " - " + df["Incentive"],
    df["Response_Rate"]
)

plt.title("Donor Response by Campaign and Incentive")
plt.xlabel("Experimental Combination")
plt.ylabel("Response Rate (%)")
plt.xticks(rotation=45)
plt.show()
```
