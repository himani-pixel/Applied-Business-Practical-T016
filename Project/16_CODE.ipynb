```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

df = pd.read_csv("blood_donation_registry_ml_ready.csv")

X = df[["lifetime_donation_count"]]
Y = df["donation_propensity_score"]

model = LinearRegression()
model.fit(X, Y)

Y_pred = model.predict(X)

print("REGRESSION ANALYSIS")

print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)

print("R2 Score:", r2_score(Y, Y_pred))

mse = mean_squared_error(Y, Y_pred)
rmse = mse ** 0.5

print("MSE:", mse)
print("RMSE:", rmse)

print("\nRegression Equation:")
print("Donation Propensity Score =",
      model.intercept_, "+",
      model.coef_[0], "* Lifetime Donation Count")

plt.scatter(X, Y)

plt.plot(X, Y_pred)

plt.title("Lifetime Donation Count vs Donation Propensity Score")
plt.xlabel("Lifetime Donation Count")
plt.ylabel("Donation Propensity Score")

plt.show()
```
