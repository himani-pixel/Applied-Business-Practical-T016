```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

days = 100
capacity = 50

# Simulate donor arrivals
low_demand = np.random.poisson(30, days)
normal_demand = np.random.poisson(50, days)
high_demand = np.random.poisson(70, days)

# Calculate average demand
print("SIMULATION MODELING")

print("Low Demand Average:", np.mean(low_demand))
print("Normal Demand Average:", np.mean(normal_demand))
print("High Demand Average:", np.mean(high_demand))

# Calculate days exceeding capacity
low_exceed = np.sum(low_demand > capacity)
normal_exceed = np.sum(normal_demand > capacity)
high_exceed = np.sum(high_demand > capacity)

print("\nDays Exceeding Capacity")
print("Low Demand:", low_exceed)
print("Normal Demand:", normal_exceed)
print("High Demand:", high_exceed)

# Plot simulation results
plt.plot(low_demand, label="Low Demand")
plt.plot(normal_demand, label="Normal Demand")
plt.plot(high_demand, label="High Demand")

plt.axhline(
    capacity,
    linestyle="--",
    label="Capacity"
)

plt.title("Simulated Daily Donor Demand")
plt.xlabel("Day")
plt.ylabel("Number of Donors")
plt.legend()
plt.show()
```
