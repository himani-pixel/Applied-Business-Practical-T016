```python
# Decision Making Under Uncertainty

standard_high = 100
standard_low = 40

targeted_high = 120
targeted_low = 30

p_high = 0.6
p_low = 0.4

standard_ev = (p_high * standard_high) + (p_low * standard_low)
targeted_ev = (p_high * targeted_high) + (p_low * targeted_low)

print("DECISION MAKING UNDER UNCERTAINTY")

print("Standard Campaign EV:", standard_ev)
print("Targeted Campaign EV:", targeted_ev)

print("\nDecision Tree Values")
print("Standard Campaign -> High Response:", standard_high)
print("Standard Campaign -> Low Response:", standard_low)
print("Targeted Campaign -> High Response:", targeted_high)
print("Targeted Campaign -> Low Response:", targeted_low)
```
