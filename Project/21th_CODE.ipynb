```python
import numpy as np
from scipy.stats import chi2_contingency
import matplotlib.pyplot as plt

# Simulated A/B Testing Data
np.random.seed(42)

group_A = np.random.binomial(1, 0.35, 1000)
group_B = np.random.binomial(1, 0.40, 1000)

conversions_A = group_A.sum()
conversions_B = group_B.sum()

total_A = len(group_A)
total_B = len(group_B)

rate_A = conversions_A / total_A
rate_B = conversions_B / total_B

print("A/B TESTING")

print("Campaign A Conversion Rate:", rate_A)
print("Campaign B Conversion Rate:", rate_B)

# Contingency Table
table = [
    [conversions_A, total_A - conversions_A],
    [conversions_B, total_B - conversions_B]
]

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to Reject H0")

# Conversion Rate Chart
plt.bar(["Campaign A", "Campaign B"], [rate_A, rate_B])
plt.title("A/B Testing Conversion Rates")
plt.xlabel("Campaign")
plt.ylabel("Conversion Rate")
plt.show()
```
