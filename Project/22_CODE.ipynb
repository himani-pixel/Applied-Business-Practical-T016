```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("blood_donation_registry_ml_ready.csv")

age = df["age"].dropna().values

np.random.seed(42)

bootstrap_medians = []

# 1000 Bootstrap Samples
for i in range(1000):
    sample = np.random.choice(
        age,
        size=len(age),
        replace=True
    )

    bootstrap_medians.append(np.median(sample))

# Bootstrap Median
bootstrap_medians = np.array(bootstrap_medians)

# 95% Confidence Interval
lower = np.percentile(bootstrap_medians, 2.5)
upper = np.percentile(bootstrap_medians, 97.5)

print("BOOTSTRAPPING")
print("Original Median Age:", np.median(age))
print("Bootstrap Median:", np.mean(bootstrap_medians))
print("95% Confidence Interval:", lower, "to", upper)

# Bootstrap Distribution
plt.hist(bootstrap_medians, bins=20)
plt.title("Bootstrap Distribution of Median Age")
plt.xlabel("Median Age")
plt.ylabel("Frequency")
plt.show()
```
