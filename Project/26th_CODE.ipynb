```python
import pandas as pd
from scipy.stats import f_oneway

# Load Dataset
df = pd.read_csv("blood_donation_registry_ml_ready.csv")

# Create Age Groups
df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 34, 49, 69],
    labels=["Young", "Middle", "Senior"]
)

# Create Groups
young = df[df["age_group"] == "Young"]["lifetime_donation_count"]
middle = df[df["age_group"] == "Middle"]["lifetime_donation_count"]
senior = df[df["age_group"] == "Senior"]["lifetime_donation_count"]

# One-Way ANOVA
f_statistic, p_value = f_oneway(
    young,
    middle,
    senior
)

print("ANOVA")

print("Young Mean:", young.mean())
print("Middle Mean:", middle.mean())
print("Senior Mean:", senior.mean())

print("F-Statistic:", f_statistic)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```
